```python
# Function to apply IQR method to fix outliers
def fix_outliers_iqr(df):
# target (CO(GT)) and the time columns are not touched
 num = df.select_dtypes(include='number').columns.drop(['CO(GT)','Hour','Month','DayOfWeek'])
 for col in num:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR
# Replace outliers with the respective bounds
 df[col] = df[col].clip(
 lower=lower_bound,
 upper=upper_bound
)
 return df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import (
mean_absolute_error,
mean_squared_error,
r2_score
)
RANDOM_STATE = 42
sns.set_theme(style="whitegrid")
```

```python
# Select columns for IQR capping
boxplot_columns = ['PT08.S1(CO)', 'C6H6(GT)', 'NOx(GT)', 'NO2(GT)']

# Apply IQR capping
for column in boxplot_columns:
    if column in df.columns:
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        df[column] = df[column].clip(
            lower=lower_bound,
            upper=upper_bound
        )

# Plot boxplots without bubbles/outlier points
plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df[boxplot_columns],
    showfliers=False
)

plt.title("Boxplots After IQR Capping")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
X,
y,
test_size=0.20,
random_state=RANDOM_STATE
)
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("\nTraining target mean / std:", round(y_train.mean(), 3), "/", round(y_train.std(), 3))
print("Testing target mean / std:", round(y_test.mean(), 3), "/", round(y_test.std(), 3))
```

```python
numeric_features = X_train.select_dtypes(include="number").columns.tolist()
categorical_features = [
column for column in categorical_columns if column in X_train.columns
]
# Remove categorical columns from the numerical feature list.
numeric_features = [
column for column in numeric_features
if column not in categorical_features
]
# Missing sensor values (-200 -> NaN) are filled with the median, then scaled
numeric_transformer = Pipeline(steps=[
("imputer", SimpleImputer(strategy="median")),
("scaler", StandardScaler())
])
categorical_transformer = Pipeline(steps=[
("imputer", SimpleImputer(strategy="most_frequent")),
("onehot", OneHotEncoder(handle_unknown="ignore"))
])
preprocessor = ColumnTransformer(
transformers=[
("numeric", numeric_transformer, numeric_features),
("categorical", categorical_transformer, categorical_features)
],
remainder="drop"
)
print("Numerical features:", numeric_features)
print("Categorical features:", categorical_features)
```

```python
# Missing values left in the training features, before preprocessing
X_train.isnull().sum()
```

```python
# Fit on the training set only, then apply the same steps to the test set
X_train_scaled = preprocessor.fit_transform(X_train)
X_test_scaled = preprocessor.transform(X_test)
print("Train shape:", X_train_scaled.shape)
print("Test shape:", X_test_scaled.shape)
print("Missing values after preprocessing:", int(np.isnan(X_train_scaled).sum() + np.isnan(X_test_scaled).sum()))
```

```python
# Feature selection: keep the 8 best features (fitted on the training set only)
selector = SelectKBest(score_func=f_regression, k=8)
X_new = selector.fit_transform(X_train_scaled, y_train)
```

```python
selected_features = np.array(numeric_features)[selector.get_support()]
print("Selected Features:")
print(selected_features)
```

```python
feature_scores = pd.DataFrame({
'Feature': numeric_features,
'Score': selector.scores_,
'Selected': selector.get_support()
})
feature_scores = feature_scores.sort_values(
by='Score',
ascending=False
)
print(feature_scores)
```

```python
# Selecting Algorithm
models = {
"Linear Regression": LinearRegression(),
"Decision Tree": DecisionTreeRegressor(random_state=RANDOM_STATE),
"Random Forest": RandomForestRegressor(n_estimators=100, random_state=RANDOM_STATE),
}
```

```python
#Store results
scores = []
for name, model in models.items():
 model.fit(X_train_scaled, y_train)
 y_pred = model.predict(X_test_scaled)
 mae = mean_absolute_error(y_test, y_pred)
 rmse = np.sqrt(mean_squared_error(y_test, y_pred))
 r2 = r2_score(y_test, y_pred)
 scores.append({
 "Model": name,
 "MAE": round(mae, 3),
 "RMSE": round(rmse, 3),
 "R2": round(r2, 3)
})
# Convert to DataFrame for nice display
results_df = pd.DataFrame(scores)
print(results_df)
```

```python
# Best model by R2: actual vs predicted
best_name = results_df.sort_values("R2", ascending=False).iloc[0]["Model"]
best_model = models[best_name]
y_pred_best = best_model.predict(X_test_scaled)
print("Best model:", best_name)
print("R2 :", round(r2_score(y_test, y_pred_best), 3))
print("MAE :", round(mean_absolute_error(y_test, y_pred_best), 3), "mg/m3")
plt.figure(figsize=(6, 6))
plt.scatter(y_test, y_pred_best, alpha=0.4)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color="red")
plt.title(f"Actual vs Predicted CO - {best_name}")
plt.xlabel("Actual CO (mg/m3)")
plt.ylabel("Predicted CO (mg/m3)")
plt.tight_layout()
plt.show()
```

```python
# Which features matter most to the Random Forest?
importances = pd.Series(
models["Random Forest"].feature_importances_,
index=numeric_features
).sort_values()
plt.figure(figsize=(8, 6))
importances.plot(kind="barh")
plt.title("Random Forest Feature Importance")
plt.tight_layout()
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, confusion_matrix, classification_report)
```

```python
# Target: 1 = high CO (above 3 mg/m3), 0 = normal CO
df["high_CO"] = (df["CO_GT"] > 3).astype(int)
```

```python
# CO_GT is removed from the predictors because the target is built from it
X = df.drop(columns=["CO_GT", "high_CO"])
y = df["high_CO"]
print(y.value_counts())

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
```

```python
# Fill missing sensor values with the median, then scale (fitted on training data only)
prep = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
X_train_scaled = prep.fit_transform(X_train)
X_test_scaled = prep.transform(X_test)

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
}

scores = []
for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)
    scores.append({
        "Model": name,
        "accuracy": round(accuracy_score(y_test, y_pred), 2),
        "precision": round(precision_score(y_test, y_pred), 2),
        "f1": round(f1_score(y_test, y_pred), 2),
        "recall": round(recall_score(y_test, y_pred), 2)
    })

results_df = pd.DataFrame(scores)
print(results_df)
```

```python
# Detail for the best model (highest F1)
best_name = results_df.sort_values("f1", ascending=False).iloc[0]["Model"]
best_pred = models[best_name].predict(X_test_scaled)
print("\nBest model:", best_name)
print(classification_report(y_test, best_pred, target_names=["Normal CO", "High CO"]))
print(confusion_matrix(y_test, best_pred))
```
